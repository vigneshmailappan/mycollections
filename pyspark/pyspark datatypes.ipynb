```python
 
from pyspark.sql.types import *
from pyspark.sql.functions import * 
from datetime import datetime  

schema = StructType([
    StructField("string_col", StringType(), True),                          # "hello"
    StructField("int_col", IntegerType(), True),                            # 42
    StructField("float_col", FloatType(), True),                            # 3.14
    StructField("double_col", DoubleType(), True),                          # 2.71828
    StructField("boolean_col", BooleanType(), True),                        # True
    StructField("long_col", LongType(), True),                              # 1234567890123
    StructField("short_col", ShortType(), True),                            # 32767
    StructField("byte_col", ByteType(), True),                              # 127
    StructField("binary_col", BinaryType(), True),                          # bytearray(b"binary")
    StructField("date_col", DateType(), True),                              # "2025-09-23"
    StructField("timestamp_col", TimestampType(), True),                    # "2025-09-23 11:00:00"
    StructField("decimal_col", DecimalType(10, 2), True),                   # 12345.67
    StructField("array_col", ArrayType(StringType()), True),                # ["a", "b", "c"]
    StructField("map_col", MapType(StringType(), IntegerType()), True),     # {"key1": 1, "key2": 2}
    StructField("struct_col", StructType([
        StructField("nested_str", StringType(), True),                      # "value"
        StructField("nested_int", IntegerType(), True)                      # 99
    ]), True)
])

 
# Sample data
data = [(
    "hello",                                    # StringType()
    42,                                         # IntegerType()
    3.14,                                       # FloatType()
    2.71828,                                    # DoubleType()
    True,                                       # BooleanType()
    1234567890123,                              # LongType()
    32767,                                      # ShortType()
    127,                                        # ByteType()
    bytearray(b"binary"),                       # BinaryType()
    datetime.strptime("2025-09-23",'%Y-%m-%d'), # DateType()
    datetime.strptime("2025-09-23 11:00:00", '%Y-%m-%d %H:%M:%S'),  # TimestampType()
    Decimal('12345.67'),                        # DecimalType(10, 2)
    ["a", "b", "c"],                            # ArrayType(StringType())
    {"key1": 1, "key2": 2},                     # MapType(StringType(), IntegerType())
    {"nested_str": "value", "nested_int": 99}   # StructType
)]


# Create DataFrame
df = spark.createDataFrame(data, schema=schema)

# Show DataFrame
df.show(truncate=False)
#df.printSchema()
```
